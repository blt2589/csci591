# Project 3
CALM (Circumpolar Active Layer Monitoring Program)

## Hypothesis:

Despite identical spatial extents and measurement densities, CALM sites will exhibit significantly levels of spatial autocorrelation in active layer thickness (ALT), reflecting site-specific environmental processes (e.g., vegetation, regional variations) that influence spatial structure.

## Data
7 CALM sites in Arctic Alaska; ALT measurements with associated metadata (region, elevation, OLT (organic layer thickness))

This dataset is pre-cleaned from a previous analysis.


**Image:** Imnavait Creek site

![Imnavait grid](img/imnavait_grid.png)

## Data & Design

Purpose:
To assess how surface soil moisture from SMAP (Soil Moisture Active Passive) satellite data corresponds with active layer thickness (ALT) at long-term monitoring sites and whether spatial patterns of soil moisture correlate with spatial autocorrelation patterns of ALT.

Primary Questions:
- Does SMAP-derived soil moisture explain spatial or temporal variability in ALT?
- Are there significant spatial clusters in soil moisture that correspond with LISA patterns in ALT?


### Datasets: 
Temporal Range: 2015–2025 (earliest SMAP data)

Cleaned CSV of active layer thickness (ALT) per site and year
- Source: CALM Program

SMAP Enhanced L3 Radiometer Global Daily 9km (SPL3SMP_E v006)
- Source: NASA NSIDC via earthaccess





### Workflow:

CALM data is cleaned and prepared

**Acquire SMAP files**
- Python script to download datasets from earthaccess library covering only summer months (June–August) and only tiles intersecting CALM sites based on centroid buffer


#### Processing:

**For each .h5 granule:**
- Extract soil_moisture dataset
- Find pixel nearest to each site centroid
- Extract date and metadata

**Aggregation**
- Group daily SMAP values by site_id and year
- Filter for June–August (in case my download script doesn't work as expected)
- Calculate seasonal mean soil moisture per site/year

**Join with CALM Data**
- Merge seasonal SMAP moisture with ALT data:
- Match on site_id and year
- Create final analysis dataset with covariates:
    - alt_depth_cm, soil_moisture, organic_layer_thickness, elevation, landform, etc.


#### Analysis:
Spatial Autocorrelation
- Global Moran’s I (ALT and soil moisture)
- Local Indicators of Spatial Association (LISA):
    - Map LISA cluster types: HH, LL, HL, LH
    - Compare ALT and SMAP spatial clusters per year
